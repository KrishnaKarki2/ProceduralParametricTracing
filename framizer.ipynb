```python
# convert mp4 to pngs

# Upload an mp4 to your frames folder
# Run this after Cells 1-2 and before Cell 3


import os
import cv2

FRAME_DIR  = '/content/drive/MyDrive/frames'
OUTPUT_EXT = 'png'   # you can also change this to jpg

# supported video extensions
VIDEO_EXTS = ('.mp4', '.mov', '.avi', '.mkv', '.webm')

# check if directory exists
if not os.path.exists(FRAME_DIR):
    print(f"error: directory '{FRAME_DIR}' does not exist. make sure your google drive is mounted.")
else:
    all_files = os.listdir(FRAME_DIR)

    # find video file
    video_files = [f for f in all_files if f.lower().endswith(VIDEO_EXTS)]

    if not video_files:
        print(f"error: no video file found in '{FRAME_DIR}'.")
        print(f"files currently in folder: {all_files}")
    else:
        video_name = video_files[0]
        VIDEO_PATH = os.path.join(FRAME_DIR, video_name)
        print(f"detected video file: {video_name}")
        print(f"path: {VIDEO_PATH}")

        # open and extract frames from video
        cap = cv2.VideoCapture(VIDEO_PATH)

        if not cap.isOpened():
            print(f"error: could not open '{video_name}'.")
        else:
            total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
            print(f"extracting approx. {total_frames} frames...")

            count = 1
            while True:
                ret, frame = cap.read()
                if not ret:
                    break

                frame_filename = os.path.join(FRAME_DIR, f"frame{count}.{OUTPUT_EXT}")
                cv2.imwrite(frame_filename, frame)
                count += 1

                if count % 100 == 0 or count == total_frames:
                    print(f"--> extracted {count}/{total_frames} frames")

            cap.release()
            print(f"\ndone successfully extracted {count - 1} frames to '{FRAME_DIR}'.")
```
